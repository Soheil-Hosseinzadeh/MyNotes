````python
import os
from dotenv import load_dotenv
from groq import Groq

# os.environ["HTTP_PROXY"] = "http://127.0.0.1:1080"
# os.environ["HTTPS_PROXY"] = "http://127.0.0.1:1080"
````

````python
def get_response(prompt: str, model: str = "openai/gpt-oss-20b") -> str:
    load_dotenv(".env")
    client = Groq(
        api_key=os.environ.get("GROQ_API_KEY"),
    )
    chat_completion = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": prompt,
            }
        ],
        model="openai/gpt-oss-20b",
    )
    return chat_completion.choices[0].message.content
````

## Principles of Prompting
#### 1. Write clear and specific instructions
#### 2. Give the model time to think

````python
prompt = """What's going on right now in the world between IRAN and the USA? answer very short and brief."""
response = get_response(prompt=prompt)
print(response)
````

## Chatbot
We can utilize the chat format to have extended conversation with chatbots personlized or specialized for specific tasks or behaviors.





````python
def get_completion_from_messages(messages: list, model: str = "openai/gpt-oss-20b") -> str:
    load_dotenv(".env")
    client = Groq(
        api_key=os.environ.get("GROQ_API_KEY"),
    )
    chat_completion = client.chat.completions.create(
        messages=messages,
        model="openai/gpt-oss-20b",
        temperature=0
    )
    return chat_completion.choices[0].message.content
````

````python
user_message = "I feel like I don't do enough, I feel as much as I try hard it just doesn't work out, maybe I shouldn't be in such a rush, maybe I should just continue, I don't know. I feel discouraged."
messages = [
    {"role": "system", "content": "You are an emotional supportive assistant that has to talk to people who has got depression and help them to find their own way. don't use markdown notations, instead just write the context with normal text format. don't make the response too long!"},
    {"role": "user", "content": user_message}
]

response = get_completion_from_messages(messages)
print(response)
````
